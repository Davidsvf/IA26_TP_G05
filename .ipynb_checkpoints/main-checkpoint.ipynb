## 1. Introdução

Este projeto aborda um problema de recolha e entrega de produtos locais.

Os produtores disponibilizam produtos, os clientes fazem encomendas e os veículos que operam a partir de centros de recolha são responsáveis por recolher e entregar esses produtos.

O objetivo é desenvolver um agente inteligente capaz de gerar rotas eficientes, respeitando várias restrições, nomeadamente:

- janelas temporais de recolha;
- janelas temporais de entrega;
- capacidade dos veículos;
- obrigatoriedade de recolher antes de entregar;
- distâncias e tempos de viagem.

Numa fase posterior serão aplicadas e comparadas diferentes técnicas de Inteligência Artificial.

## 2. Resumo da Análise Exploratória dos Dados

O dataset pequeno será analisado no notebook `01_eda_tiny.ipynb`.

A análise irá incluir:

- estrutura dos dados;
- valores em falta e valores duplicados;
- produtores, produtos, clientes e encomendas;
- janelas temporais de recolha e entrega;
- capacidades dos veículos;
- localizações utilizadas nas rotas;
- distâncias e tempos de viagem.

O objetivo desta análise é compreender corretamente os dados antes da implementação dos algoritmos de pesquisa.

## 3. Definição do Problema

O problema consiste em encontrar rotas válidas e eficientes para os veículos disponíveis.

Cada rota deverá:

1. recolher os produtos nos respetivos produtores;
2. respeitar as janelas temporais de recolha;
3. respeitar a capacidade máxima do veículo;
4. entregar os produtos nos respetivos locais de entrega;
5. respeitar as janelas temporais de entrega;
6. garantir que cada produto é recolhido antes de ser entregue.

O objetivo é encontrar soluções viáveis e, ao mesmo tempo, reduzir o custo das rotas.

## 4. Abordagem Proposta

O projeto será desenvolvido por etapas.

Numa primeira fase será realizada a análise dos dados e a modelação do problema.

De seguida será implementado um mecanismo para verificar se uma rota é válida, tendo em conta:

- janelas temporais;
- capacidade dos veículos;
- ordem entre recolha e entrega;
- distâncias e tempos de viagem.

Posteriormente serão aplicadas diferentes técnicas de Inteligência Artificial:

- Greedy Best-First Search;
- A* Search;
- Hill-Climbing;
- Constraint Satisfaction Problem (CSP).

Por fim, os resultados obtidos pelos diferentes métodos serão comparados.
